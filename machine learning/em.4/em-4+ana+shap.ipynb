```python
# 获取单个样本的Top N特征值和对应的SHAP值
def get_topN_reason(old_list, features, top_num = 3, min_value = 0.0):
  # 输出shap值最高的N个标签
  feature_importance_dict = {}
  for i, f in zip(old_list, features):
    feature_importance_dict[f] = i
  new_dict = dict(sorted(feature_importance_dict.items(), key=lambda e: e[1], reverse=True))
  return_dict = {}
  for k ,v in new_dict.items():
    if top_num > 0:
      if v >= min_value:
        return_dict[k] = v
        top_num -= 1
      else:
        break
    else:
      break
  return return_dict
feat_name=['Ar','VE','AR ','EN','Ea',' IE1 ','CoE']   #  0 相对原子质量 1价电子 2原子半径 3电负性 4电子亲和能 5第一电离能 6内聚能
print(get_topN_reason(old_list=shap_values[23], features=feat_name)) 
```
