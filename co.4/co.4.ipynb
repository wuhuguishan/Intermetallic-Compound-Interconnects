```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import RepeatedKFold
from modules import *
import sys
import math
import numpy as np
import pandas as pd
from scipy.stats import pearsonr
from sklearn.feature_selection import SelectKBest
from scipy.stats import pearsonr
import seaborn as sns   

def corre_spearman(X,label,y=None,mask_typ=True,output=True,fname='feat_corr', plt_prop=True):
  score_matrix0=[]
  score_matrix=[]

  x = pd.DataFrame(X)
  score_matrix0=x.corr("spearman")
  score_matrix0[score_matrix0<0.01]=0
  score_matrix = score_matrix0.values

  plt.rcParams['figure.facecolor'] = 'none'  # 画布透明
  plt.rcParams['axes.facecolor'] = 'none'    # 坐标轴透明
  fig, ax = plt.subplots(figsize=(8, 8))
    
  mask = np.zeros_like(score_matrix, dtype=np.bool_)
  mask[np.triu_indices_from(mask,k=1)] = True
 
  
  d1=pd.DataFrame(score_matrix,index=label,columns=label)
  cmap = sns.cubehelix_palette(start = 1.5, rot = 3, gamma= 0.8, hue=0.8,as_cmap = True)
  if mask_typ :
     sh=sns.heatmap(d1,vmin=-1,vmax=1,square=True, center=True,mask=~mask,annot=True,cmap=cmap,cbar=False,annot_kws={"size":10},ax=ax)
  else:
     sh=sns.heatmap(d1,vmin=-1,vmax=1,square=True, linewidths=1.5,center=True,annot=True,cmap=cmap,cbar=False,annot_kws={"size":10})
     

  sh.set_xticklabels(sh.get_xticklabels(),rotation=90,fontsize=12)
  sh.set_yticklabels(sh.get_yticklabels(),fontsize=12,va='center')
  sh.xaxis.set_ticks_position('top')  
  sh.xaxis.set_label_position('top')
  sh.yaxis.set_ticks_position('right')
  sh.yaxis.set_label_position('right')
  ### colorbar sets ###
  cb=sh.figure.colorbar(sh.collections[0],fraction=0.04, pad=0.1)
  cb.ax.tick_params(labelsize=12)
  cb.set_label('spearman correlation coffecient',fontsize=15)
  
  if output:
     plt.savefig(f'{fname}.png',dpi=300,bbox_inches='tight',transparent=True)
  if plt_prop:
     plt.savefig('spearman.png',dpi=300,bbox_inches='tight')
     plt.show()
     
  return score_matrix

File = '../data.4/co-wt'
X,y= sub_data(File)
X,y= pre_data(X,y)
X=X[:55,:]
y=y[:55]
print(X.shape)
print(y.shape)
label=['Ar','VE','AR ','EN','Ea',' IE1 ','BP'] 
corre_spearman(X,label)
```

```python
# 获取单个样本的Top N特征值和对应的SHAP值
def get_topN_reason(old_list, features, top_num = 3, min_value = 0.0):
  # 输出shap值最高的N个标签
  feature_importance_dict = {}
  for i, f in zip(old_list, features):
    feature_importance_dict[f] = i
  new_dict = dict(sorted(feature_importance_dict.items(), key=lambda e: e[1], reverse=True))
  return_dict = {}
  for k ,v in new_dict.items():
    if top_num > 0:
      if v >= min_value:
        return_dict[k] = v
        top_num -= 1
      else:
        break
    else:
      break
  return return_dict
feat_name=['Ar','VE','AR ','EN','Ea',' IE1 ','BP']   #  0 相对原子质量 1价电子 2原子半径 3电负性 4电子亲和能 5第一电离能 6内聚能
print(get_topN_reason(old_list=shap_values[23], features=feat_name)) 
```

```python
import pandas as pd
import matplotlib.pyplot as plt  
feat_name=['Ar','VE','AR ','EN','Ea',' IE1 ','BP']
X_all = pd.DataFrame(X2, columns=feat_name)
shap_values = explainer.shap_values(X_all) # 传入特征矩阵X，计算SHAP值
# summarize the effects of all the features

fig = plt.figure()
shap.summary_plot(shap_values, X_all,cmap = "plasma",show=False )
plt.gcf().set_size_inches(9,6)
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.xlabel("SHAP value",fontsize=20)
plt.savefig('summary_plot_co.png', dpi=300, bbox_inches='tight')
plt.savefig('summary_plot_co.svg', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
# summarize the feature importance
fig = plt.figure()
shap.summary_plot(shap_values, X_all, plot_type="bar",color='orange',show = False)
# shap.summary_plot(shap_values, X_all, plot_type="bar",show = False)
shap_df = pd.DataFrame(shap_values, columns=X_all.columns)
feature_importance = shap_df.abs().mean().sort_values(ascending=False)
print(feature_importance.head())

plt.gcf().set_size_inches(9,6)
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.xlabel("mean(|SHAP value|)",fontsize=20)
plt.savefig('summary_plot_value_co.png', dpi=300, bbox_inches='tight')
plt.savefig('summary_plot_value_co.svg', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```
